####Load Documents

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_data = TextLoader("../data/speech.txt",encoding="utf-8")
pdf_data = PyPDFLoader("../data/medical_report.pdf")

text = text_data.load()
pdf = pdf_data.load()
```

###Split documents

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=20)
```

```python
text_split = splitter.split_documents(documents=text)
pdf_split = splitter.split_documents(documents=pdf)

print(len(text_split))
print(len(pdf_split))
```
