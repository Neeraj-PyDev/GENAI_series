```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = ["Machine learning is a subset of artificial intelligent.",
             "Deep Learning uses neural networks with multiple layers.",
             "Python is a programming language popular for data science.",
             "India won the cricket world cup on 2011.",
             "AI models can understand and generate human language."]
```

```python
from langchain_openai.embeddings import OpenAIEmbeddings

emb = OpenAIEmbeddings(model="text-embedding-3-small")

vector = emb.embed_documents(documents) #manual

print(len(vector[4]))
```

```python
from langchain_chroma import Chroma
vector_store = Chroma.from_texts(
    texts=documents,
    embedding=emb,
    persist_directory="./vector"
)
```

###Retrieve results from our document/Vector db

```python
query = "Data Science & ML"
res = vector_store.similarity_search(query=query, k=2)
res
```
