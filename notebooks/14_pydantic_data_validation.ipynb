```python
from pydantic import BaseModel, Field
class Product(BaseModel):
    title:str = Field(max_length=20)
    expiry:str
    desc:str = Field(min_length=20)
```

```python
product1 = Product(title="Nokia",expiry="5 years",desc="This is Nokia 23 series with brand new features..")
product2 = Product(title="Samsung",expiry="4 years",desc="This is samsung S23,new mobile launched in 2026..")
```

```python
from typing import List

class carts(BaseModel):
    cart_id:str
    items:List[Product]
    comment:str = Field(default="No comment.")
```

```python
my_cart = carts(cart_id="sjk1234",items=[product1,product2])

cart = [item.title for item in my_cart.items ]

print(cart)
```

```python
## Annotated

from typing import Annotated

common_validator = Annotated[List[Product], Field(default=[])]
```

```python
class Item(BaseModel):
    item_id:str 
    price:int = Field(default=100)
    items:common_validator
```

```python
item1 = Item(item_id="ejue12",price=23)

item1
```
