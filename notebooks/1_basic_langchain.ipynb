#### Connect with OpenAI model

```python
import os
from dotenv import load_dotenv

load_dotenv()

from langchain_openai import ChatOpenAI

llm_oai = ChatOpenAI(model = "gpt-4")

resp = llm_oai.invoke("Who is PM of India?")

print(resp.content)
```

#### Connect with Gemini

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm_gen = ChatGoogleGenerativeAI(model = "gemini-3.1-pro-preview")
query = "Who is best batsman of India?"
resp = llm_gen.invoke(query)
# print(resp.content)
```
