```python
from langchain_openai import ChatOpenAI
llm_oai = ChatOpenAI(model = "gpt-4o")
```

```python
prompts = [
    {"role":"system", "content":"you are a Javascript developer."},
    {"role":"user", "content":"difference b/w map and filter."}
]

res = llm_oai.invoke(prompts)
print(res.content)
```

#### Prompt Template - Dynamic Prompts

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages ([
    {"role":"system", "content":"you are a translator to translate in {language}."},
    {"role":"user", "content":"{query}"}
])
```

```python
final_prompts = prompts.format_messages(language="Hinglish", query="I want to learn Javascript.")
resp = llm_oai.invoke(final_prompts)
print(resp.content)
```

#### Chaining Prompts

```python
resp_2 = llm_oai.invoke(prompts.invoke({"language":"hinglish","query":"I love Langchain."}))
print(resp_2.content)
```

```python
# We can do same thing above in few words like below:

chains = prompts | llm_oai

resp_3 = chains.invoke({"language":"hinglish","query":"I love Langchain."})
print(resp_3.content)
```

```python
# Now we will use one more pipe. So that we dont need to write content.

chain_1 = prompts | llm_oai | out
resp_4 = chain_1.invoke({"language":"hindi","query":"I love Langchain."})
print(resp_4)
```

#### My self chain

```python
#function

def my_chain(result : str):
    return result.upper()
```

```python
my_chain = prompts | llm_oai | out | my_chain
resp_5 = my_chain.invoke({"language":"hindi","query":"I love Langchain"})
print(resp_5)
```
