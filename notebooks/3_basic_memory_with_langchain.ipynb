#### Memory management

```python
history=[]
while True:
    query= input("User: ")
    if query.lower() in ['quit','bye','exit']:
        print('Goodbye!! 👋')
        break

    history.append({"role":"user", "content": query})
    print("User: ", query)

    res = llm_oai.invoke(history)
    history.append({"role":"ai", "content": res.content})
    print("AI: ", res.content, '\n')


```
