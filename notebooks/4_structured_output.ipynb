```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model= 'gpt-4o')
```

#### Problem Statement

```python
text="My name is Neeraj Kumar, my email is neeraj27@gmail.com and age is 32."
```

```python
res= llm.invoke(f"Please extract only name,email and age from this text:{text}")
print(res.content)
```

#### Solution is: Structured Output.

```python
from pydantic import BaseModel, Field
from typing import List

class structred_output(BaseModel):
    name: str = Field(description= "Full name")
    email: str = Field(description= "email")
    age: int = Field(description= "Age")

structred_llm = llm.with_structured_output(structred_output)

resp = structred_llm.invoke(f"Please extract only name,email and age from this text:{text}")

```

```python
resp.model_dump()
```

#### Movie Recommend

```python
class movies(BaseModel):
    title: str = Field(description="Movie title")
    year: int = Field(description="Release date")

movie_llm = llm.with_structured_output(movies)

res = movie_llm.invoke("suggest a sci-fi movie.")

res
```

```python
## to get more than one movies.

class allmovies(BaseModel):
    movie: List[movies]

movies_llm = llm.with_structured_output(allmovies)

reslt = movies_llm.invoke("suggest some hindi horror movies?")

print(reslt)
```
