#### streaming openai chats

```python
from langchain_openai import ChatOpenAI

llm_oai = ChatOpenAI(model = 'gpt-4o', streaming=True)

prompts = [{"role":"user", "content":"Who is Rohit sharma?"}]

res = llm_oai.stream(prompts)

for i in res:
    print(i.content, end="")
```

#### streaming groq models

```python
from langchain_groq import ChatGroq

llm_qroq = ChatGroq(model = "openai/gpt-oss-120b", streaming= True)

query = "Explain Himalayas in details."

res = llm_qroq.stream(query) # this become iterable object.

for chunk in res:
    print(chunk.content, end="")
```
