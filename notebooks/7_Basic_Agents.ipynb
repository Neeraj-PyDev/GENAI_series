```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain.tools import tool

llm_oai = ChatOpenAI(model = "gpt-4o")
```

#### Defining Tools

```python
@tool
def add_num(a:int, b:int)-> int:
    """
    It will add two numbers.
    Args
        a:number1
        b:number2
    """
    return a+b

@tool
def mul_num(a:int, b:int)-> int:
    """
    It will multiply two numbers.
    Args
        a:number1
        b:number2
    """
    return a*b
```

```python
add_num.invoke({'a':2, 'b':3})
```

#### Creating Agents

```python
agent = create_agent(
    model=llm_oai,
    tools= [add_num, mul_num],
    system_prompt= "This is a number agent,which does calculations."
)

resp = agent.invoke({'messages':[{'role':'user', 'content':'what is 239 add 390'}]})

print(resp)
```

#### REtrieve output from Agent

```python
print(resp['messages'][-1].content)
```
