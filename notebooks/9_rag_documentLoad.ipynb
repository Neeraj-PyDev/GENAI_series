#### Text doc load

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader(
    "../data/speech.txt",
    encoding="utf-8"
)

docs = loader.load()
len(docs)

onedoc = docs[0]
print(onedoc.page_content)
```

#### PDF Load

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_loader = PyPDFLoader("../data/medical_report.pdf")

pdf = pdf_loader.load()

len(pdf)

pdf_firstpage = pdf[8].page_content

print(pdf_firstpage)
```
